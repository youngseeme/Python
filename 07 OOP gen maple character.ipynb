```python
import random
```

```python
class Character:
    def __init__(self, name, STR, LUK, DEX, INT):
        self.name = name
        self.STR = STR
        self.LUK = LUK
        self.DEX = DEX
        self.INT = INT

    @staticmethod
    def roll_dice():
        return random.randint(4, 8), random.randint(4, 8), random.randint(4, 8), random.randint(4, 8)

    def show_info(self):
        print(f"\n 캐릭터명: {self.name}")
        print(f" STR: {self.STR}, LUK: {self.LUK}, DEX: {self.DEX}, INT: {self.INT}")
        
```

```python
characters = []


while True:
    print('============메이플 캐릭터 만들기============')
    print('1) 캐릭터 생성\n2) 캐릭터 조회\n3) 프로그램 종료 ')

    choice = int(input('번호를 입력하세요: '))

    if choice == 1:
        name = input('캐릭터 명을 입력하세요: ').strip()

        if any(c.name == name for c in characters):
            print("이미 있는 캐릭터 이름입니다!")
            continue

        while True:
            print('주사위를 굴립니다!')
            STR, LUK, DEX, INT = Character.roll_dice()
            print(f'결과) STR: {STR}, LUK: {LUK}, DEX: {DEX}, INT: {INT}')
            again = input('재시도 하시겠습니까? (y/n)').upper
            if again == 'Y':
                continue
            else:
                break

        new_char = Character(name, STR, LUK, DEX, INT)
        characters.append(new_char)
        print(f"\n캐릭터 '{name}' 생성 완료!")

        
    elif choice == 2:
        print("\n캐릭터 목록:")
        if not characters:
            print("아직 생성된 캐릭터가 없습니다.")
        else:
            for c in characters:
                c.show_info()

    elif choice == 3:
        print('프로그램을 종료합니다.')
        break
                
    else:
        print('잘못 입력하셨습니다.')
```
