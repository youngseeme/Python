```python
import numpy as np
```

### **실습 1: 칼로리 소모량 점수 계산하기 🔥**

AI 모델이 `[운동 시간(분), 평균 심박수(BPM), 체중(lb)]` 3가지 요소로 운동 중 소모된 칼로리 점수를 예측하는 시나리오입니다.

#### **📜 개념 및 사용법**
가중합은 각 데이터 값에 해당하는 가중치를 곱한 후, 그 결과들을 모두 더하여 계산합니다. 이 계산은 NumPy의 내적(Dot Product) 함수 `np.dot()`을 사용하면 매우 빠르고 간결하게 구현할 수 있습니다.

> **계산 공식**: `(데이터1 × 가중치1) + (데이터2 × 가중치2) + ...`

```python
# 예제: 사용자 A의 칼로리 소모량 점수 계산

# 사용자 A의 운동 데이터: [운동 시간: 30분, 평균 BPM: 120, 체중: 150lb]
user_A = np.array([30, 120, 150])

# 가설 1. '운동 시간'이 가장 중요하다고 보는 모델의 가중치
weights_v1 = np.array([5.5, 1.2, 0.8])

# np.dot() 함수로 가중합 계산
calorie_score_A_v1 = np.dot(user_A, weights_v1)

print(f"가설 1 모델로 계산한 사용자 A의 칼로리 점수: {calorie_score_A_v1:.2f}")
```

#### **✍️ 실습 문제 1-1: 가중치를 바꾸어 결과 관찰하기**

새로운 연구 결과, **'체중'과 '심박수'가 칼로리 소모에 더 결정적인 요소**라는 가설이 제기되었습니다. 이 가설에 따라 가중치를 다음과 같이 조정했습니다.

* **가설 2 (새로운 가중치)**: `[운동 시간: 3.0, 평균 BPM: 1.8, 체중: 1.5]`

**사용자 A** (`[30, 120, 150]`)의 칼로리 소모량 점수가 이 새로운 가중치에 따라 어떻게 변하는지 `np.dot()`을 사용하여 계산하고, **결과가 어떻게 달라지는지 관찰**해 보세요.

```python
weights_v2 = np.array([3.0, 1.8, 1.5])
calorie_score_A_v2 = np.dot(user_A, weights_v2)
print('가설 2 모델로 계산한 사용자 A의 칼로리 점수: ', calorie_score_A_v2)
```

### **실습 2: 심장질환 발병 위험도 점수 계산하기 ❤️**

이번에는 다른 AI 모델이 `[총 콜레스테롤, 수축기 혈압, 일일 평균 걸음수]` 3가지 요소로 심장질환 위험도를 예측하는 시나리오입니다.

#### **✍️ 실습 문제 2-1: 새로운 모델로 위험도 계산**

* **모델 가중치**: `[콜레스테롤: 0.8, 혈압: 0.4, 걸음수: -0.002]`
    *(참고: 걸음수 가중치가 음수인 이유는, 걸음수가 많을수록 위험도는 낮아지기 때문입니다.)*

* **환자 B의 데이터**: `[총 콜레스테롤: 240, 수축기 혈압: 145, 일일 평균 걸음수: 7500]`

환자 B의 심장질환 위험도 점수를 계산해 보세요.

```python
user_B = [240, 145, 7500]
heart_weights = [0.8, 0.4, -0.002]
heart_disease_B = np.dot(user_B, heart_weights)
print('환자 B의 심장질환 위험도 점수: ', heart_disease_B)
```

---
### **실습 3 (심화): 여러 환자의 위험도 한번에 계산하기 🏥**

#### **📜 개념 및 사용법**
실제 AI 모델은 수많은 환자 데이터를 한번에 처리합니다. NumPy의 `np.dot()`은 **행렬(여러 환자 데이터)과 벡터(가중치) 간의 연산**도 지원하여, 반복문 없이 모든 환자의 점수를 한번에 계산할 수 있습니다.

```python
# 예제: 여러 환자의 당뇨병 위험도 점수 계산

# 3명의 환자 데이터를 행렬로 정의: [혈당, 체중, 나이]
patient_matrix = np.array([
    [150, 92, 58], # 환자 A
    [95, 60, 35],  # 환자 C
    [180, 110, 62]  # 환자 D
])

# 당뇨병 모델 가중치
diabetes_weights = np.array([0.7, 0.2, 0.1])

# 행렬과 벡터의 내적으로 모든 환자의 점수를 한번에 계산!
all_scores = np.dot(patient_matrix, diabetes_weights)

print("모든 환자의 당뇨병 위험도 점수:")
print(f"환자 A: {all_scores[0]:.2f}")
print(f"환자 C: {all_scores[1]:.2f}")
print(f"환자 D: {all_scores[2]:.2f}")
```

#### **✍️ 실습 문제 3-1: 여러 사용자의 칼로리 소모량 계산**

아래 3명의 사용자 데이터 행렬(`user_matrix`)과 위에서 사용한 **'가설 2'** 칼로리 모델 가중치(`weights_v2`)를 이용해, **모든 사용자의 칼로리 소모량 점수를 한 줄의 코드로 계산**하고 결과를 출력하세요.

```python
# 3명의 사용자 데이터: [운동 시간(분), 평균 BPM, 체중(lb)]
user_matrix = np.array([
    [30, 120, 150], # 사용자 A
    [60, 140, 180], # 사용자 B
    [45, 110, 130]   # 사용자 C
])

# '가설 2' 칼로리 모델 가중치
weights_v2 = np.array([3.0, 1.8, 1.5])
```

```python
all_c = np.dot(user_matrix, weights_v2)

print(f'칼로리 소모량 점수\nA: {all_c[0]:.2f}\nB: {all_c[1]}\nC: {all_c[2]}')
```
