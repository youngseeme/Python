# 실습과제 1: 사칙연산 미니 계산기

---


## 🎯 목표

- 사용자로부터 숫자를 입력받아 사칙연산 계산기를 만들어봅니다.

- 입력값을 int()로 정수형으로 변환합니다.

- 산술 연산자(+, -, *, /)를 사용합니다.

- 조건문(if / elif / else)으로 어떤 연산을 수행할지 분기합니다.

- f-string을 활용해 결과를 보기 좋게 출력합니다.

```python
# TODO: 사용자로부터 첫 번째 숫자를 입력받으세요
print('* 미니 사칙연산기 *')
print('아래 안내에 따라 정수와 연산기호를 입력해주세요')
print("  ")
num1 = int(input("첫 번째 숫자를 입력하세요: "))
# TODO: 사용자로부터 두 번째 숫자를 입력받으세요
num2 = int(input("두 번째 숫자를 입력하세요: "))
# TODO: 연산자(+,-,*,/)를 입력받으세요
operator = input("+, -, *, / 중 한 개의 연산기호를 입력하세요: ")
# TODO: 조건문(if/elif/else)으로 연산을 수행하세요
if operator == '+':
    result = num1 + num2
elif operator == '-':
    result = num1 - num2
elif operator == '*':
    result = num1 * num2
elif operator == '/':
    if num2 != 0:
        result = num1 / num2
    else:
        result = "0으로는 나눌 수 없습니다."
else:
    result = "잘못된 수식입니다."
print("  ")
print(f'계산 결과는 [ {result} ]입니다.')

# TODO: f-string을 사용하여 결과를 출력하세요
```

# 실습과제 2: 온도 단위 변환기

---


## 🎯 목표

- 사용자로부터 길이와 온도를 입력받아 단위를 변환해봅니다.

- 센티미터 → 미터로 길이를 변환합니다.

- **섭씨(℃) → 화씨(℉)**로 온도를 변환합니다.

- 입력값은 float()로 실수형으로 변환합니다.

- 변환 결과는 f-string을 이용해 보기 좋게 출력합니다.

```python
# TODO: 사용자로부터 cm 값을 입력받으세요
print('* cm를 m로 단위 변환하기 *')
print('아래 안내에 따라 변환할 길이(cm)를 입력해주세요')
print("  ")
cm = float(input("cm : "))
# TODO: cm → m 변환 기능을 작성하세요
m = cm * 0.01
# TODO: 변환 결과를 출력하세요
print("  ")
print(f"{cm}cm는 {m}m 입니다.")
# TODO: 사용자로부터 섭씨(℃) 값을 입력받으세요
print(" ")
print(" ")
print("* 섭씨(C)를 화씨(F)로 변환하기 *")
print('아래 안내에 따라 변환할 온도(섭씨)를 입력해주세요')
print("  ")
# TODO: 섭씨 → 화씨 변환 기능을 작성하세요
celsius = float(input("섭씨(C): "))
fahrenheit = 9/5 * celsius + 32
# TODO: 변환 결과를 출력하세요
print("  ")
print(f'섭씨(C) {celsius} 도는 화씨(F) {fahrenheit} 도 입니다.')
```
